# AI Workshop: Decision Trees to XGBoost

## Clinical AI Learning Path — Module 2 Overview

This module takes you from a single decision tree through to XGBoost, using three clinical
prediction problems that build on one another. It is the foundation for the XGBoost deep
dive in Module 3.

---

## 🎯 Learning Objectives

By the end of this module you will be able to:

1. **Read** a decision tree the way a clinician reads a protocol, and explain any single prediction
2. **Recognize** why a single tree is unstable, and what that instability costs in a clinical setting
3. **Apply** Random Forest to noisy multi-channel monitoring data
4. **Use** Gradient Boosting where interactions between patient factors matter
5. **Move** to XGBoost for production-grade tabular prediction

## 🏥 Why these three problems

Each was chosen because it exercises a different property of tree models.

**Sepsis triage** is the case for a single tree. The decision has to be explainable in the
moment, to a colleague or to a patient, and a shallow tree is one of the few models that can
be read directly. The section also shows the cost of that readability: retrain on a slightly
different sample and the splits move.

**Ward escalation** is the case for a forest. Twelve monitoring channels, each noisy on its
own, none decisive alone. Averaging across trees is what makes the prediction stable, and the
feature importances tell you which channels are actually earning their place on the chart.

**Length of stay** is the case for boosting. The drivers interact: severity compounds with
age, frailty compounds with the number of procedures, and discharge delays produce a long
right tail that a linear model cannot follow.

## 📊 Expected Outcomes

**Technical skills**
- Build and read decision trees for clinical classification
- Build Random Forest models for deterioration scoring
- Apply Gradient Boosting to length of stay and capacity prediction
- Understand what XGBoost adds over plain gradient boosting

**Operational understanding**
- Estimate the value of earlier deterioration detection
- Use length of stay forecasts for bed capacity planning
- Judge when interpretability outweighs accuracy, and when it does not

**Judgement**
- Choose between a tree, a forest and a boosted ensemble for a given problem
- Recognize class imbalance and why accuracy is the wrong measure for it
- Anticipate distribution shift as case mix changes

## ⚠️ Before you start

**About the data.** Every patient in this module is simulated from published physiological
reference ranges. No real or identifiable patient data is used anywhere in this course.

**About the models.** Nothing here is a validated clinical tool. These are teaching examples
built to show how the algorithms behave. A model intended for real clinical use needs
prospective validation, subgroup fairness analysis and regulatory review.
